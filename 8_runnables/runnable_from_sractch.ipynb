```python
# Cell 3 - NakliPromptTemplate
class NakliPromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
# Cell 4 - Without chains
template = NakliPromptTemplate(
    template='Write a {length} blog post about {topic}',
    input_variables={'length', 'topic'}
)
template.format({'length': 'short', 'topic': 'Machine Learning'})
```

```python
# Cell 5 - NakliLLMChain
class NakliLLMChain:
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self, input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)
        return result['response']
```

```python
# Cell 6 - Using chains
template = NakliPromptTemplate(
    template='Write a {length} blog post about {topic}',
    input_variables={'length', 'topic'}
)

llm = NakliLLM()
chain = NakliLLMChain(llm, template)
chain.run({'length': 'detailed', 'topic': 'Machine Learning'})
```

```python
# Cell 7 - Runnable abstract class
from abc import ABC, abstractmethod

class Runnable(ABC):
    @abstractmethod
    def invoke(self, input_data):
        pass
```

```python
# Cell 8 - NakliLLM with Runnable
import random

class NakliLLM(Runnable):
    def __init__(self):
        print('LLM created successfully')

    def invoke(self, prompt):
        response_list = [
            'Python is the most popular programming language for AI',
            'Neural networks are inspired by the human brain',
            'Data is the new oil in the digital world',
            'Deep learning has revolutionized computer vision and NLP'
        ]
        return {'response': random.choice(response_list)}

    def predict(self, prompt):
        response_list = [
            'Python is the most popular programming language for AI',
            'Neural networks are inspired by the human brain',
            'Data is the new oil in the digital world',
            'Deep learning has revolutionized computer vision and NLP'
        ]
        return {'response': random.choice(response_list)}
```

```python
# Cell 9 - NakliPromptTemplate with Runnable
class NakliPromptTemplate(Runnable):
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def invoke(self, input_dict):
        return self.template.format(**input_dict)

    def format(self, input_dict):
        return self.template.format(**input_dict)
```

```python
# Cell 10 - NakliStrOutputParser
class NakliStrOutputParser(Runnable):
    def __init__(self):
        pass

    def invoke(self, input_data):
        return input_data['response']
```

```python
# Cell 10 - NakliStrOutputParser
class NakliStrOutputParser(Runnable):
    def __init__(self):
        pass

    def invoke(self, input_data):
        return input_data['response']
```

```python
# Cell 11 - RunnableConnector
class RunnableConnector(Runnable):
    def __init__(self, runnable_list):
        self.runnable_list = runnable_list

    def invoke(self, input_data):
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
        return input_data
```

```python
# Cell 12 - parser
parser = NakliStrOutputParser()
```

```python
# Cell 13 - Single chain test
llm = NakliLLM()
template = NakliPromptTemplate(
    template="Write a {length} summary about {topic}",
    input_variables=["length", "topic"]
)
chain = RunnableConnector([template, llm, parser])
chain.invoke({'length': 'brief', 'topic': 'Artificial Intelligence'})
```

```python
# Cell 14 - template1
template1 = NakliPromptTemplate(
    template="Write a fun fact about {topic}",
    input_variables=["topic"]
)
```

```python
# Cell 15 - template2
template2 = NakliPromptTemplate(
    template="Summarize the following in simple words: {response}",
    input_variables=["response"]
)
```

```python
# Cell 16 - Final chained output
llm = NakliLLM()
parser = NakliStrOutputParser()
chain1 = RunnableConnector([template1, llm])
chain2 = RunnableConnector([template2, llm, parser])
final_chain = RunnableConnector([chain1, chain2])
final_chain.invoke({'topic': 'Deep Learning'})
```
